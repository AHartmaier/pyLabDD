```python
import numpy as np
import matplotlib.pyplot as plt
import pylabdd as dd
```

```python
temp = 900.0
tau0 = 0.0
lgb = 0.1
grain_size = 1.0
maxout = 1000
tfin = 8.e8
niter = 50_000_000
ngbn = 25 #int(lgb/delta)
delta = lgb/(ngbn - 1)  # 0.004
dtmax = 6.25e-7/delta**4  # value 1.e4 for delta=0.005
gbdd = dd.GB_dislocations(temp=temp, tau0=tau0,
                   Ngbn=ngbn, len_gb_seg=lgb, grain_size=grain_size,
                   tfin=tfin, niter=niter, maxout=maxout,
                   dtmax=dtmax,
                   screenout=True)
gbdd.run_sim()
```

```python
#gbdd.plot_time_series(['timestep', 'n_gbdis_eff'])
gbdd.plot_field('bfield', last_frame=200, nplot=5, fs=26, lw=4, file="gb01_t900_zoom.pdf")
gbdd.plot_field('bfield', last_frame=900, nplot=6, fs=26, lw=4, file="gb01_t900_full.pdf")
gbdd.plot_field('displacement', last_frame=900, nplot=6, lw=4, fs=26, file="gb01_t900.pdf")
gbdd.plot_field('flux', last_frame=900, nplot=6, fs=26, lw=4, file="gb01_t900.pdf")
```

```python
print(f"** Max. time step = {max(gbdd.glob_data['timestep'])}, Median: {np.median(gbdd.glob_data['timestep'])}")
```
